Реляційна алгебра визначає комплекс операцій з відношеннями (таблицями): поєднання, різницю, перетин тощо.

Всі ці операції можуть бути виражені через мову SQL, що ми й будемо з вами робити.

Терміни:
- сутність — те, що ми описуємо (будь-який конкретний товар, якась конкретна людина, замовлення тощо);
- таблиця (відношення) — це множина будь-яких сутностей;
- стовпчик таблиці (атрибут) — один з параметрів якими ми описуємо сутності;
- рядок таблиці (запис, кортеж) — набір атрибутів з конкретними значеннями;
- фінальний набір — результат запиту SQL. Завжди таблиця (навіть коли результат — просте число).

Таблиці (стовпці — атрибути, рядки — сутності (дані), шапка таблиці — імена атрибутів).

### Коротко про SQL

SQL (від англ. Structured Query Language — «мова структурованих запитів» — спеціалізована декларована мова програмування.

Мова SQL стандартизована. Діючий стандарт — ANSI SQL-92. Як правило, різні БД підтримують цей стандарт та додають свої розширення (свій діалект).

Складається з декількох груп операторів:
- DDL (Data Definition Language) — CREATE, DROP, ALTER;
- DML (Data Manipulation Language) — SELECT, INSERT, UPDATE, DELETE;
- TCL (Transaction Control Language) — COMMIT, ROLLBACK, SAVEPOINT;
- DCL (Data Control Language) — GRANT, REVOKE, DENY.

### Використання вбудованої бібліотеки Python sqlite3 для роботи з БД SQLite.

SQLite, мабуть, є найпростішою базою даних для підключення до програми Python, оскільки для цього не потрібно встановлювати жодних зовнішніх модулів Python SQL. За замовчуванням ваша інсталяція Python містить бібліотеку Python SQL під назвою sqlite3, яку можна використовувати для взаємодії з базою даних SQLite.

Більше того, бази даних SQLite є автономними та безсерверними, оскільки вони читають та записують дані у файл. Це означає, що, на відміну від MySQL або PostgreSQL, навіть не потрібно встановлювати та запускати сервер SQLite для виконання операцій з базою даних!

Підключимось до БД:

````python
import sqlite3
from sqlite3 import Error


def create_connection(path):
    connection = None
    try:
        connection = sqlite3.connect(path)
        print("Connection to SQLite DB successful")
    except Error as e:
        print(f"The error '{e}' occurred")

    return connection
````

Ось як працює цей код:

Рядки 1 й 2 імпортують модуль та клас Error з модуля sqlite3.
Рядок 5 визначає функцію .create_connection(), яка приймає у якості параметра шлях до бази даних SQLite.
Рядок 8 використовує метод .connect() з модуля sqlite3 та приймає шлях до бази даних SQLite в якості параметру. Якщо база даних існує у вказаному місці, тоді встановлюється з’єднання з базою даних, інакше — у вказаному місці створюється нова база даних та встановлюється з’єднання.
Рядок 9 друкує статус успішного підключення до бази даних.
Рядок 10 відловлює будь-які винятки, які можуть бути згенеровані, якщо методу .connect() не вдається встановити з'єднання.
Рядок 11 друкує повідомлення про помилку у консолі.

Метод .connect(path) з модуля sqlite3 повертає об’єкт connection, який, у свою чергу, повертає функція create_connection() на 13 рядку. Цей об’єкт connection можна використовувати для виконання запитів до бази даних SQLite. Наступний сценарій створює підключення до бази даних SQLite:

````python
connection = create_connection("./data/sqliteDB")
````

Після виконання наведеного вище сценарію можна побачити, що файл бази даних sqliteDB створено у відповідному каталозі, якщо до цього його не існувало.

Треба звернути увагу, що можна змінити розташування відповідно до ваших налаштувань.

### Створення таблиць

Щоб виконувати запити в SQLite, використовуйте метод .execute() об'єкту cursor. У цьому розділі визначимо функцію execute_query(), яка використовує цей метод. Функція прийматиме об’єкт connection та рядок запиту, який буде надано у метод .execute() об'єкту cursor.

connection.execute() може виконувати будь-який запит, наданий йому у вигляді рядка. Використаємо цей метод для створення таблиць у цьому розділі. У наступних розділах використовуватимемо цей метод для виконання запитів на оновлення та видалення.

[Докладніше про типи даних в SQLite](https://www.sqlite.org/datatype3.html).

[Докладніше про команду CREATE TABLE](https://www.sqlite.org/lang_createtable.html).



````python
def execute_query(connection, query):
    cursor = connection.cursor()
    try:
        cursor.execute(query)
        connection.commit()
        print("Query executed successfully")
    except Error as e:
        print(f"The error '{e}' occurred")
````

Цей код намагається виконати задане query та друкує повідомлення про помилку, якщо необхідно.

Далі пишемо запит:

````python
create_users_table = """
CREATE TABLE IF NOT EXISTS users (
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  name TEXT NOT NULL,
  age INTEGER,
  gender TEXT,
  nationality TEXT
);
"""
````

Це говорить про створення таблиці users з такими п’ятьма стовпцями:

- id;
- name;
- age;
- gender;
- nationality.

Викличемо функцію execute_query() для створення таблиці та надаємо об’єкт connection, створений у попередньому розділі, разом із рядковим об'єктом create_users_table, який містить запит на створення таблиці:

````python
execute_query(connection, create_users_table)
````

Таблиця posts:

````python
create_posts_table = """
CREATE TABLE IF NOT EXISTS posts(
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  title TEXT NOT NULL,
  description TEXT NOT NULL,
  user_id INTEGER NOT NULL,
  FOREIGN KEY (user_id) REFERENCES users (id)
);
"""
````

Оскільки між таблицями users та posts відношення "один-до-багатьох", створюємо вторинний ключ user_id у таблиці posts, який пов'язаний з колонкою id таблиці users.

````python
execute_query(connection, create_posts_table)
````

Аналогічно створюємо таблиці comments, likes, emails:

````python
create_comments_table = """
CREATE TABLE IF NOT EXISTS comments (
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  text TEXT NOT NULL,
  user_id INTEGER NOT NULL,
  post_id INTEGER NOT NULL,
  FOREIGN KEY (user_id) REFERENCES users (id) FOREIGN KEY (post_id) REFERENCES posts (id)
);
"""

create_likes_table = """
CREATE TABLE IF NOT EXISTS likes (
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  user_id INTEGER NOT NULL,
  post_id integer NOT NULL,
  FOREIGN KEY (user_id) REFERENCES users (id) FOREIGN KEY (post_id) REFERENCES posts (id)
);
"""

create_emails_table = """
CREATE TABLE IF NOT EXISTS emails (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER NOT NULL,
    email TEXT NOT NULL,
    FOREIGN KEY (user_id) REFERENCES users (id)
);
"""

execute_query(connection, create_comments_table)
execute_query(connection, create_likes_table)
execute_query(connection, create_emails_table)
````

### Додавання записів

Щоб додати записи до бази даних SQLite, можна скористатися тією ж функцією execute_query(), яку використовували для створення таблиць. По-перше, треба зберегти свій INSERT INTO запит у рядку. Потім можна передати об’єкт connection та рядок query у execute_query().

##### Наприклад:

insert_users = """
INSERT INTO
    users (name, age, gender, nationality)
VALUES
    (name_1, age_1, gender_1, country_1),
    (name_2, age_2, gender_2, country_2),
    ...
    (name_n, age_n, gender_n, country_n);
"""

Замість name_1, age_1, gender_1 і nationality_1 підставляємо конкретні значення атрибутів та надаємо рядок для виконання у створену раніше функцію execute_query() - для виконання нашого запиту.

Так само робимо й з іншими таблицями.
Важливо(!): усі дані повинні бути узгоджені згідно обмежень (вторинні ключі (FOREGIN KEY) та обов'язковість параметрів (NOT NULL)).

Розберемо на прикладі:


Для створення фейкових даних для наших таблиць використаємо пакет Faker — [посилання на документацію й приклади використання тут](https://faker.readthedocs.io/en/master/).

Тепер необхідні фейкові дані можуть створюватися просто:

````python
print(fake.name())
print(fake.name_female())
print(fake.name_male())
print(fake.country())
print(fake.email())
print(fake.sentence(nb_words=6, variable_nb_words=True))
print(fake.paragraph(nb_sentences=4, variable_nb_sentences=True))
````

Для генерації віку людини та створення випадково чоловіків та жінок використаємо можливості вбудованого модулю random:

````python
import random
print(f"age: {random.randint(10, 90)}")
gender = "female" if random.choice((True, False)) else "male"
print(gender)
````

Заповнимо таблиці users, emails та posts.
Генерацію постів до таблиці posts для кожного користувача будемо робити на основі випадкової вибірки:

````python
def user_gen(num: int) -> str:
    step = 0
    while True:
        genger = "female" if random.choice((True, False)) else "male"
        name = fake.name_male() if genger == "male" else fake.name_female()
        age = random.randint(10, 90)
        nationality = fake.country()
        yield f"""("{name}", {age}, "{genger}", "{nationality}")"""
        step += 1
        if step >= num:
            break


insert_users = """
    INSERT INTO
        users (name, age, gender, nationality)
    VALUES
""" + ",\n".join(list(user_gen(300))) + ";"
print(insert_users)
````

Тепер можна створити цю таблицю:

````python
execute_query(connection, insert_users)
````

Для створення інших таблиць нам необхідне значення поля id для кожного конкретного значення запису, бо воно присутнє у якості вторинного ключа. У реальному житті, коли ви додаєте користувачів по одному, ви отримуєте й це значення. При створенні фейкової таблиці, після створення таблиці користувачів просто будемо читати її зачення та на базі їх — створювати інші таблиці.

Зробимо деякі допущення:
- для таблиці emails: у людини може бути від одного нуля до трьох email, причому один буває вчетверо частіше ніж жодного, чи три, та вдвічі частіше ніх три (це просто допущення, так би мовити — "зі стелі"). Кількість email для генерації буде обиратись для кожного користувача за допомогою функції random.choice((0, 1, 1, 1, 1, 2, 2, 3));
- кількість постів для конкретної людини будемо випадково обирати random.choice((0, 1, 2, 2, 3, 3, 3, 4, 4, 5, 6, 7)).

Потрібно:
- прочитати усі існуючі users.id;
- ітеруватися по усім існуючим users.id та створити рядки для додавання записів у таблиці emails та posts (аналогічно створеним для вставки користувачів);
- коли рядки будуть створені — передати їх на виконання.

#### Вибір записів

Щоб обрати записи за допомогою SQLite, ми можемо знову використати метод .execute() об'єкту cursor. Однак після того, як це зробити, треба буде викликати метод .fetchall(). Цей метод повертає список кортежів, де кожен кортеж зіставляється з відповідним рядком в отриманих записах.

[Докладніше про fetchall() і інші методи об'єкта _cursor_ ](https://docs.python.org/3/library/sqlite3.html?highlight=fetchall#sqlite3.Cursor.fetchall)

Щоб спростити процес, можна створити функцію execute_read_query():

````python
def execute_read_query(connection, query):
    cursor = connection.cursor()
    result = None
    try:
        cursor.execute(query)
        result = cursor.fetchall()
        return result
    except Error as e:
        print(f"The error '{e}' occurred")
````

Тепер оберемо усі записи з таблиці users:
(Не рекомендується використовувати SELECT * для великих таблиць, оскільки це може призвести до великої кількості операцій вводу/виводу, що збільшує мережевий трафік):

````python
select_users = "SELECT * from users"
users = execute_read_query(connection, select_users)

for user in users:
    print(user)
````

Для формування інших таблиць потрібні тільки id, тому сформуємо запит, який обере тільки їх:

````python
select_users_id = "SELECT id from users;"
users_id = execute_read_query(connection, select_users_id)

for id in users_id:
    print(id)
````

Тепер можна створити таблиці emails та posts:

````python
email_list = []
post_list = []
for id_user in users_id:
    email_list.extend([f"({id_user[0]}, '{fake.email()}')" for _ in range(random.choice((0, 1, 1, 1, 1, 2, 2, 3)))])
    post_list.extend([
            f"('{fake.sentence(nb_words=6, variable_nb_words=True)}', 
               '{fake.paragraph(nb_sentences=4, variable_nb_sentences=True)}', {id_user[0]})" 
                for _ in range(random.choice((0, 1, 2, 2, 3, 3, 3, 4, 4, 5, 6, 7)))
    ])

print(email_list)
print(post_list)
````

Тепер, коли у нас є інформація для додавання у таблиці, створимо необхідні рядки:

````python
insert_emails = """
    INSERT INTO
        emails (user_id, email)
    VALUES
""" + ",\n".join(email_list) + ";"

insert_posts = """
    INSERT INTO
        posts (title, description, user_id)
    VALUES
""" + ",\n".join(post_list) + ";"


print(insert_emails)
print(insert_posts)
````

````python
execute_query(connection, insert_emails)
execute_query(connection, insert_posts)
````

Тепер, коли у нас є повні таблиці users та posts (отже, ми можемо отримати users.id та posts.id, які входять до таблиці likes та comments, в якості вторинних ключів — FOREIGN KEY). Все аналогічно попередній ситуації, але якщо потрібні первинні ключі (PRIMARY KEY) не тільки для users, але й для posts.

Вважатимемо, що для кожного поста може бути:

likes_num = random.choice((2, 3, 4, 4, 5, 5, 6, 7, 8, 9, 10, 11, 12)) лайків та

comments_num = random.choice((0, 1, 2, 2, 3, 4)) коментарі.

Для того, щоб обрати користувачів, які зробили відповідну кількість лайків або постів, будемо робити:

users_who_liked = random.choices(users_id, likes_num)

users_who_commented = random.choices(users_id, comments_num)


Отримаємо список всіх posts_id (прочитаємо аналогічно users_id):

````python
select_posts_id = "SELECT id from posts;"
posts_id = execute_read_query(connection, select_posts_id)

for id in posts_id:
    print(id)
````

Тепер можна створювати пости та лайки ітеруючись по списку posts_id:

````python
comments_list = []
likes_list = []
for post_id in posts_id:
    # отримаємо кількість коментарів під постом
    comments_num = random.choice((0, 1, 2, 2, 3, 4))
    # отримаємо список користувачів, які створювали ці коментарі
    users_who_commented = random.choices(users_id, k=comments_num)
    # створюємо список кортежів для додавання рядка інформації у БД
    for i in range(comments_num):
        comments_list.append(f"('{fake.sentence(nb_words=6, variable_nb_words=True)}', {users_who_commented[i][0]}, {post_id[0]})")

    # отримаємо кількість лайків під постом
    likes_num = random.choice((2, 3, 4, 4, 5, 5, 6, 7, 8, 9, 10, 11, 12))
    # отримаємо список користувачів, які лайкали
    users_who_liked = random.choices(users_id, k=likes_num)
    # створюємо список кортежів для додавання рядка інформації у БД
    for i in range(likes_num):
        likes_list.append(f"({users_who_liked[i][0]}, {post_id[0]})")

print(comments_list)
print(likes_list)
````

````python
insert_comments = """
    INSERT INTO
        comments (text, user_id, post_id)
    VALUES
""" + ",\n".join(comments_list) + ";"

insert_likes = """
    INSERT INTO
        likes (user_id, post_id)
    VALUES
""" + ",\n".join(likes_list) + ";"

print(insert_comments)
print(insert_likes)
````

````python
execute_query(connection, insert_comments)
execute_query(connection, insert_likes)
````

#### DISTINCT

Оператор використовується для вибору унікальних рядків, наприклад, якщо треба мати список країн, у яких живуть зараз наші користувачі.

ВАЖЛИВО! Якщо обираємо декілька атрибутів, то у вибірку потраплять лише унікальні комбінації:

````python
distinct_query = """
SELECT DISTINCT nationality
FROM users;
"""
````

````python
distinct_result = execute_read_query(connection, distinct_query)
print(distinct_result)
````

#### COUNT

Оператор використовується для підрахунку кількості отриманих у виборці рядків.

Важливо! COUNT лише рахує рядки. Якщо необхідно підрахувати унікальні рядки — необхідно комбінувати з DISTINCT.


````python
count_query = """
SELECT COUNT(DISTINCT nationality)
FROM users;
"""
````

````python
count_result = execute_read_query(connection, count_query)
print(count_result)
````

#### WHERE

Оператор використовується для фільтрації даних за умовою:

Формат:
```SQL
SELECT <*>
FROM <table_name>
WHERE <condition>;
```

condition ---> boolean result

У фінальну вибірку потрапляють рядки, для яких умова буде мати значення true:

Наприклад, можливі умови (можуть бути й інші):
- a = b;
- a > b;
- a >= b;
- a < b;
- a <= b;
- a != b.

````python
where_query = """
SELECT id, name, age
FROM users
WHERE age > 40;
"""
````

````python
where_result = execute_read_query(connection, where_query)
print(where_result)
````

Усе, що вивчили раніше, також можна використовувати.
Сформуємо запит, який поверне кількість людей з віком більше за 40 років:

````python
where_count_query = """
SELECT COUNT(*)
FROM users
WHERE age > 40;
"""
````

````python
where_count_result = execute_read_query(connection, where_count_query)
print(where_count_result)
````
